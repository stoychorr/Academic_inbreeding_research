```python
import re
import time
import pandas as pd
from urllib.parse import urljoin

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import (
    StaleElementReferenceException,
    TimeoutException,
    ElementClickInterceptedException
)
from webdriver_manager.chrome import ChromeDriverManager


BASE_URL = "https://ras.nacid.bg/"


def get_text_safe(elem):
    try:
        return elem.text.strip()
    except:
        return ""


def extract_link_from_row(driver, row):
    """
    Tries:
    1. direct href
    2. onclick/js
    3. clicking the icon and capturing resulting URL
    """
    # 1) direct href
    try:
        anchors = row.find_elements(By.TAG_NAME, "a")
        for a in anchors:
            href = a.get_attribute("href")
            if href and href.strip():
                return urljoin(BASE_URL, href)
    except:
        pass

    # 2) onclick/js
    try:
        candidates = row.find_elements(By.CSS_SELECTOR, "a, button, input, span, i, svg")
        for el in candidates:
            onclick = el.get_attribute("onclick")
            if onclick:
                m = re.search(r"""['"]([^'"]+)['"]""", onclick)
                if m:
                    candidate = m.group(1)
                    if "/" in candidate or "http" in candidate:
                        return urljoin(BASE_URL, candidate)
    except:
        pass

    # 3) fallback click
    try:
        original_window = driver.current_window_handle
        old_handles = set(driver.window_handles)
        old_url = driver.current_url

        clickable_candidates = row.find_elements(By.CSS_SELECTOR, "a, button, i, svg")
        for el in clickable_candidates:
            try:
                driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", el)
                time.sleep(0.2)
                driver.execute_script("arguments[0].click();", el)
                time.sleep(1.2)

                new_handles = set(driver.window_handles)
                if len(new_handles) > len(old_handles):
                    new_window = list(new_handles - old_handles)[0]
                    driver.switch_to.window(new_window)
                    new_url = driver.current_url
                    driver.close()
                    driver.switch_to.window(original_window)
                    return new_url

                new_url = driver.current_url
                if new_url != old_url:
                    driver.back()
                    time.sleep(1.2)
                    return new_url
            except:
                continue
    except:
        pass

    return None


def scrape_current_table(driver):
    rows_data = []
    table_rows = driver.find_elements(By.CSS_SELECTOR, "table tbody tr")

    for row in table_rows:
        try:
            cells = row.find_elements(By.TAG_NAME, "td")
            if len(cells) < 4:
                continue

            row_number = get_text_safe(cells[0]) if len(cells) > 0 else ""
            name = get_text_safe(cells[1]) if len(cells) > 1 else ""
            degree = get_text_safe(cells[2]) if len(cells) > 2 else ""
            current_position = get_text_safe(cells[3]) if len(cells) > 3 else ""
            previous_position = get_text_safe(cells[4]) if len(cells) > 4 else ""

            link = extract_link_from_row(driver, row)

            rows_data.append({
                "№": row_number,
                "Име": name,
                "Научна степен": degree,
                "Текуща академична длъжност": current_position,
                "Заемана академична длъжност": previous_position,
                "Линк": link
            })

        except StaleElementReferenceException:
            continue

    return rows_data


def wait_for_user_filters_and_table(driver, wait):
    print("\nChrome opened.")
    print("Please apply the filters manually in the browser:")
    print("  1. Open the blue tab: ВСИЧКИ ЛИЦА НА АКАД. ДЛЪЖНОСТИ И ПРИДОБИТИ НАУЧНИ СТЕПЕНИ")
    print("  2. Организация = Нов български университет")
    print("  3. Академична длъжност = Професор, Доцент, Главен асистент, Асистент")
    print("  4. Настояща = checked")
    print("  5. Заемана = unchecked")
    print("  6. Брой резултати = 100")
    print("  7. Click Търси and wait for the table to load")
    input("\nWhen the filtered results table is visible, press ENTER here... ")

    wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "table tbody tr")))
    time.sleep(1)


def get_row_count(driver):
    try:
        return len(driver.find_elements(By.CSS_SELECTOR, "table tbody tr"))
    except:
        return 0


def click_load_more_until_end(driver, wait, pause=1.5, max_clicks=100):
    """
    Clicks 'Още резултати' repeatedly until it disappears or stops adding rows.
    """
    print("\nExpanding all results...")
    click_count = 0
    unchanged_rounds = 0

    while click_count < max_clicks:
        before_count = get_row_count(driver)

        button = None
        button_xpaths = [
            "//button[contains(normalize-space(), 'Още резултати')]",
            "//a[contains(normalize-space(), 'Още резултати')]",
            "//*[contains(@class,'btn') and contains(normalize-space(), 'Още резултати')]",
        ]

        for xp in button_xpaths:
            try:
                candidates = driver.find_elements(By.XPATH, xp)
                visible = [c for c in candidates if c.is_displayed()]
                if visible:
                    button = visible[0]
                    break
            except:
                continue

        if button is None:
            print("No 'Още резултати' button found anymore.")
            break

        try:
            driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", button)
            time.sleep(0.4)
            driver.execute_script("arguments[0].click();", button)
        except ElementClickInterceptedException:
            try:
                time.sleep(0.8)
                driver.execute_script("arguments[0].click();", button)
            except:
                print("Could not click 'Още резултати'.")
                break
        except:
            print("Could not click 'Още резултати'.")
            break

        time.sleep(pause)

        # Wait a bit for row count to increase
        increased = False
        for _ in range(8):
            after_count = get_row_count(driver)
            if after_count > before_count:
                increased = True
                break
            time.sleep(0.5)

        click_count += 1

        after_count = get_row_count(driver)
        print(f"  Click {click_count}: rows {before_count} -> {after_count}")

        if after_count == before_count:
            unchanged_rounds += 1
        else:
            unchanged_rounds = 0

        # If clicking no longer changes anything twice in a row, stop
        if unchanged_rounds >= 2:
            print("Row count stopped changing. Assuming all results are loaded.")
            break

    print(f"Finished expanding after {click_count} clicks.")
    print(f"Total visible rows now: {get_row_count(driver)}")


def main():
    options = webdriver.ChromeOptions()
    # options.add_argument("--headless=new")

    driver = webdriver.Chrome(
        service=Service(ChromeDriverManager().install()),
        options=options
    )

    wait = WebDriverWait(driver, 30)

    try:
        driver.get("https://ras.nacid.bg/dissertations-search")

        wait_for_user_filters_and_table(driver, wait)

        # Click "Още резултати" until everything is loaded
        click_load_more_until_end(driver, wait)

        # Final scrape of the fully expanded table
        print("\nScraping full expanded table...")
        all_rows = scrape_current_table(driver)

        df = pd.DataFrame(all_rows)
        df = df.drop_duplicates()

        out_file = "nacid_nbu_results.xlsx"
        df.to_excel(out_file, index=False)

        print(f"\nSaved {len(df)} rows to {out_file}")

    finally:
        driver.quit()


if __name__ == "__main__":
    main()
```
